### Notebook 3 : Linear regression full_dataset

##### Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

##### Load the data set

```python
df = pd.read_csv('capstone.csv')
```

##### Prepare the data set

* Drop MemberName since it is not necessary

```python
df = df.drop(columns=['MemberName'])
```

* Map 'CourseName' to numeric

```python
df['CourseName'] = df['CourseName'].map({
    'Foundations of ML': 0,
    'Production-Ready ML Systems': 1
})
```

* Feature matrix X and target y

```python
feature_cols = ['EducationLevel', 'Attendance', 'TotalHours',
                'AssignmentsCompleted', 'HackathonParticipation',
                'GitHubScore', 'PeerReviewScore', 'CourseName']

X = df[feature_cols].values
y = df['CapstoneScore'].values

print("X shape:", X.shape)
print("y shape:", y.shape)
```

#####  Feature Scaling

```python
# Calculate mean and std deviation for each column (feature-wise)
X_mean = np.mean(X, axis=0)
X_std = np.std(X, axis=0)

# Standardize each feature manually
X_scaled = (X - X_mean) / X_std

print("X_scaled shape:", X_scaled.shape)
```

##### Multivariate Gradient Descent

```python
def compute_cost(X, y, beta):
    m = len(y)
    predictions = X @ beta
    errors = predictions - y
    cost = (1 / (2 * m)) * np.sum(errors ** 2)
    return cost

def gradient_descent(X, y, learning_rate=0.01, num_iterations=1000):
    m, n = X.shape
    # Add bias column (ones) to X
    X_bias = np.hstack((np.ones((m, 1)), X))
    # Initialize beta vector
    beta = np.zeros(n + 1)
    cost_history = []

    for i in range(num_iterations):
        predictions = X_bias @ beta
        errors = predictions - y
        gradients = (1 / m) * X_bias.T @ errors
        beta -= learning_rate * gradients

        cost = compute_cost(X_bias, y, beta)
        cost_history.append(cost)

        if i % 100 == 0 or i == num_iterations - 1:
            print(f"Iteration {i}: Cost={cost:.2f}")

    return beta, cost_history
```

```python
beta, cost_history = gradient_descent(X_scaled, y, learning_rate=0.00001, num_iterations=1000)
print("Final beta values for learning rate 0.00001 : :", beta)
```

##### Plotting Cost vs Iterations for Each Learning Rate

```python
def plot_single_cost(cost_history, alpha_value):
    plt.figure(figsize=(8, 5))
    plt.plot(range(len(cost_history)), cost_history, color='blue')
    plt.xlabel('Iterations')
    plt.ylabel('Cost (MSE)')
    plt.title(f'Cost vs Iterations (α = {alpha_value})')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
plot_single_cost(cost_history, 0.00001)
```

```python
beta, cost_history = gradient_descent(X_scaled, y, learning_rate=0.001, num_iterations=1000)
print("Final beta values for learning rate 0.001 : ", beta)
```

```python
plot_single_cost(cost_history, 0.001)
```

```python
plot_single_cost(cost_history, 0.1)
```

```python
plot_single_cost(cost_history, 1.0)
```

```python
y_mean = np.mean(y)
baseline_mse = np.mean((y - y_mean)**2)
print(f"Baseline MSE (predicting the mean): {baseline_mse:.2f}")
```

```python
def predict(X, beta):
    # Add bias column to X
    X_bias = np.c_[np.ones((X.shape[0], 1)), X]  # shape becomes (150, 9)
    return np.dot(X_bias, beta)

def r2_score(y_true, y_pred):
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    return 1 - (ss_res / ss_tot)

# Predict using the best beta
y_pred = predict(X_scaled, beta)

# Calculate R²
r2 = r2_score(y, y_pred)
print(f"R² Score: {r2:.4f}")
```

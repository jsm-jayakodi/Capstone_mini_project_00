### Notebook 5 : Sklearn Comparison

##### Import libraries

##### Load data set

```python
import pandas as pd
import numpy as np
```

##### Load dataset

```python

df = pd.read_csv(r'C:\Users\HP\Desktop\mini project 00\capstone.csv')
```

##### Data Preperation

* Drop non-numeric and irrelevant columns

```python
df = df.drop(columns=["MemberName"])
```

* Map CourseName to numeric values

```python
course_mapping = {
    'Foundations of ML': 0,
    'Production-Ready ML Systems': 1
}
df['CourseName'] = df['CourseName'].map(course_mapping)
```

* Separate features (X) and target (y)

```python
feature_cols = [
    'EducationLevel', 'Attendance', 'TotalHours',
    'AssignmentsCompleted', 'HackathonParticipation',
    'GitHubScore', 'PeerReviewScore', 'CourseName'
]
X = df[feature_cols].values
y = df['CapstoneScore'].values

print("X shape:", X.shape)
print("y shape:", y.shape)
```

* Scale the Features

```python
from sklearn.preprocessing import StandardScaler

# Initialize the scaler
scaler = StandardScaler()

# Fit and transform the feature matrix
X_scaled = scaler.fit_transform(X)

print("X_scaled shape:", X_scaled.shape)
```

##### Linear Regression using scikit-learn

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Fit scikit-learn's Linear Regression model
linreg = LinearRegression()
linreg.fit(X_scaled, y)

# Predict using the model
y_pred = linreg.predict(X_scaled)

# Evaluate performance
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

# Print results
print("Linear Regression (sklearn)")
print("Intercept (beta_0):", linreg.intercept_)
print("Coefficients (beta):", linreg.coef_)
print("Mean Squared Error:", mse)
print("R² Score:", r2)
```

##### Polynomial Regression (Degree 2) – scikit-learn

```python
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, r2_score

# Set degree
degree = 2

# Build the pipeline: scale -> polynomial expansion -> linear regression
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('poly', PolynomialFeatures(degree=degree, include_bias=True)),
    ('linreg', LinearRegression())
])

# Fit the model
pipeline.fit(X, y)

# Predict and evaluate
y_pred = pipeline.predict(X)
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"Polynomial Regression (Degree {degree})")
print("Mean Squared Error:", mse)
print("R² Score:", r2)
```

##### Cross-Validation for Degree 2 Polynomial Model

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score

# Step 1: Create pipeline
degree = 2
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('poly', PolynomialFeatures(degree=degree, include_bias=True)),
    ('linreg', LinearRegression())
])

# Step 2: Perform cross-validation (5-fold) and use R² as scoring metric
cv_scores = cross_val_score(pipeline, X, y, cv=5, scoring='r2')

# Step 3: Print the results
print(f"R² scores for each fold: {cv_scores}")
print(f"Average R² score: {np.mean(cv_scores)}")
```

##### Polynomial Regression (Degree 3) – scikit-learn

```python
# Set degree
degree = 3

# Build the pipeline: scale -> polynomial expansion -> linear regression
poly3_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('poly', PolynomialFeatures(degree=degree, include_bias=True)),
    ('linreg', LinearRegression())
])

# Fit the model
poly3_pipeline.fit(X, y)
y_pred3 = poly3_pipeline.predict(X)

mse = mean_squared_error(y, y_pred3)
r2 = r2_score(y, y_pred3)

print(f"Polynomial Regression (Degree {degree})")
print("Mean Squared Error:", mse)
print("R² Score:", r2)
```

##### Cross-Validation for Degree 3 Polynomial Model

```python
# 5-fold CV
cv_scores3 = cross_val_score(poly3_pipeline, X, y, cv=5, scoring='r2')
print("Fold R² scores (deg 3):", cv_scores3)
print("Average R² (deg 3):", np.mean(cv_scores3))
```
